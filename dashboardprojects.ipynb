```python
import sqlite3
from tkinter import *
from tkinter import messagebox, ttk, filedialog
from PIL import Image, ImageTk
import urllib.request
import io
import re

class Error404Window:
    def __init__(self, root):
        self.root = root
        self.root.title("404 Error - Access Denied")
        self.root.geometry("450x300+450+250")
        self.root.config(bg="#d9534f")
        
        Label(self.root, text="404", font=("Helvetica", 45, "bold"), bg="#d9534f", fg="white").pack(pady=10)
        Label(self.root, text="Invalid Password Format!", font=("Helvetica", 16, "bold"), bg="#d9534f", fg="white").pack()
        
        msg = "Error: Password cannot contain special characters like #, @, $, etc.\nPlease use standard alphanumeric characters only."
        Label(self.root, text=msg, font=("Helvetica", 11), bg="#d9534f", fg="#f2f2f2", justify=CENTER).pack(pady=15)
        
        Button(self.root, text="Back to Login", command=self.go_back, font=("Helvetica", 11, "bold"), bg="white", fg="#d9534f", cursor="hand2", relief=RAISED).pack(ipadx=10, ipady=5)

    def go_back(self):
        self.root.destroy()


class LoginWindow:
    def __init__(self, root):
        self.root = root
        self.root.title("Student Management System Login")
        self.root.geometry("900x550+250+100")
        self.root.config(bg="#1e1e2f")
        
        header_frame = Frame(self.root, bg="#1e1e2f", height=60)
        header_frame.pack(side=TOP, fill=X)
        
        lbl_univ = Label(header_frame, text="STUDENT MANAGEMENT SYSTEM", font=("Helvetica", 18, "bold"), bg="#1e1e2f", fg="#00adb5")
        lbl_univ.place(x=30, y=15)

        body_frame = Frame(self.root, bg="#2b2d42")
        body_frame.pack(fill=BOTH, expand=True)
        
        self.bg_canvas = Canvas(body_frame, width=900, height=490, highlightthickness=0)
        self.bg_canvas.pack(fill=BOTH, expand=True)
        
        image_urls = [
            "https://images.unsplash.com/photo-1523050854058-8df90110c9f1?w=900",
            "https://images.unsplash.com/photo-1541339907198-e08756dedf3f?w=900",
            "https://images.unsplash.com/photo-1498243691581-b145c3f54a5a?w=900"
        ]
        
        self.img_index = 0
        self.tk_images = []
        self.bg_canvas.create_rectangle(0, 0, 900, 490, fill="#1e1e2f")
        
        for url in image_urls:
            try:
                req = urllib.request.Request(url, headers={'User-Agent': 'Mozilla/5.0'})
                with urllib.request.urlopen(req, timeout=3) as response:
                    image_data = response.read()
                    pil_img = Image.open(io.BytesIO(image_data)).resize((900, 490), Image.Resampling.LANCZOS)
                    self.tk_images.append(ImageTk.PhotoImage(pil_img))
            except Exception:
                pass
                
        if self.tk_images:
            self.bg_image_item = self.bg_canvas.create_image(0, 0, anchor=NW, image=self.tk_images[0])
            self.start_slideshow()
            
        login_frame = Frame(body_frame, bg="white", bd=1, relief=SOLID)
        login_frame.place(x=520, y=80, width=330, height=310)
        
        red_bar = Frame(login_frame, bg="#b30000", height=45)
        red_bar.pack(side=TOP, fill=X)
        Label(red_bar, text="Sign in to your account", font=("Helvetica", 13, "bold"), bg="#b30000", fg="white").pack(pady=10)
        
        Label(login_frame, text="User Name", font=("Helvetica", 10, "bold"), bg="white", fg="#555").place(x=30, y=70)
        self.txt_user = Entry(login_frame, font=("Helvetica", 11), bg="#f9f9f9", bd=1, relief=SOLID)
        self.txt_user.place(x=30, y=95, width=270, height=28)
        
        Label(login_frame, text="Password", font=("Helvetica", 10, "bold"), bg="white", fg="#555").place(x=30, y=135)
        self.txt_pass = Entry(login_frame, show="*", font=("Helvetica", 11), bg="#f9f9f9", bd=1, relief=SOLID)
        self.txt_pass.place(x=30, y=160, width=270, height=28)
        
        btn_forgot = Button(login_frame, text="Forgot Password", font=("Helvetica", 9), bg="white", fg="#0066cc", bd=0, cursor="hand2")
        btn_forgot.place(x=25, y=198)
        
        btn_register = Button(login_frame, text="New Registration", font=("Helvetica", 9), bg="white", fg="#0066cc", bd=0, cursor="hand2")
        btn_register.place(x=190, y=198)
        
        btn_login = Button(login_frame, text="SIGN IN", command=self.login_action, font=("Helvetica", 11, "bold"), bg="#2196f3", fg="white", cursor="hand2", bd=0)
        btn_login.place(x=30, y=235, width=270, height=35)
        
        footer_lbl = Label(self.root, text="© 2026 Student Management System. All Rights Reserved.", font=("Helvetica", 8), bg="#1e1e2f", fg="white")
        footer_lbl.pack(side=BOTTOM, fill=X, ipady=4)

    def start_slideshow(self):
        if self.tk_images:
            self.img_index = (self.img_index + 1) % len(self.tk_images)
            try:
                self.bg_canvas.itemconfig(self.bg_image_item, image=self.tk_images[self.img_index])
                self.root.after(3500, self.start_slideshow)
            except Exception:
                pass

    def login_action(self):
        username = self.txt_user.get()
        password = self.txt_pass.get()
        
        if username == "" or password == "":
            messagebox.showerror("Error", "Please enter User Name and Password", parent=self.root)
            return
            
        if not re.match("^[a-zA-Z0-9]+$", password):
            err_root = Toplevel(self.root)
            Error404Window(err_root)
            return
            
        messagebox.showinfo("Success", f"Welcome, {username}!", parent=self.root)
        self.root.destroy()
        root2 = Tk()
        StudentResultManagementSystem(root2)
        root2.mainloop()


class StudentResultManagementSystem:
    def __init__(self, root):
        self.root = root
        self.root.title("Student Management System")
        self.root.geometry("1280x680+50+10")
        self.root.config(bg="#0F172A")
        self.root.focus_force()
        
        self.create_db()
        
        top_bar = Frame(self.root, bg="#1E293B", height=50)
        top_bar.pack(side=TOP, fill=X)
        
        Label(top_bar, text="🎓 STUDENT MANAGEMENT SYSTEM", font=("Helvetica", 15, "bold"), bg="#1E293B", fg="#38BDF8").place(x=20, y=10)
        
        btn_home = Button(top_bar, text="🏠 Logout / Back to Login", command=self.go_to_home, font=("Helvetica", 10, "bold"), bg="#EF4444", fg="white", cursor="hand2", relief=FLAT)
        btn_home.place(x=1050, y=8, width=210, height=34)
        
        style = ttk.Style()
        style.theme_use('default')
        style.configure('TNotebook', background='#0F172A', borderwidth=0)
        style.configure('TNotebook.Tab', background='#1E293B', foreground='#94A3B8', padding=[12, 6], font=('Helvetica', 10, 'bold'))
        style.map('TNotebook.Tab', background=[('selected', '#38BDF8')], foreground=[('selected', '#0F172A')])

        self.notebook = ttk.Notebook(self.root)
        self.notebook.pack(fill=BOTH, expand=1)
        
        self.tab_dashboard = Frame(self.notebook, bg="#0F172A")
        self.tab_student = Frame(self.notebook, bg="#F4F7F6")
        self.tab_result = Frame(self.notebook, bg="#F4F7F6")
        self.tab_fees = Frame(self.notebook, bg="#F4F7F6")
        self.tab_exam = Frame(self.notebook, bg="#F4F7F6")
        self.tab_attendance = Frame(self.notebook, bg="#F4F7F6")
        self.tab_assignment = Frame(self.notebook, bg="#F4F7F6")
        self.tab_leave = Frame(self.notebook, bg="#F4F7F6")
        self.tab_sports = Frame(self.notebook, bg="#F4F7F6")
        self.tab_exams_comp = Frame(self.notebook, bg="#F4F7F6")
        
        self.notebook.add(self.tab_dashboard, text=" 📊 Dashboard ")
        self.notebook.add(self.tab_student, text=" 👨‍🎓 Students ")
        self.notebook.add(self.tab_result, text=" 📜 Result ")
        self.notebook.add(self.tab_fees, text=" 💳 Fees ")
        self.notebook.add(self.tab_exam, text=" 📝 Exam Form ")
        self.notebook.add(self.tab_attendance, text=" 📅 Attendance ")
        self.notebook.add(self.tab_assignment, text=" 📂 Assignments ")
        self.notebook.add(self.tab_leave, text=" ✉️ Leave App ")
        self.notebook.add(self.tab_sports, text=" 🏆 Sports ")
        self.notebook.add(self.tab_exams_comp, text=" 🏅 Qualifying Exams ")
        
        self.bg_images = {}
        self.setup_tab_bg(self.tab_student, "https://images.unsplash.com/photo-1523240795612-9a054b0db644?w=1280", "student")
        self.setup_tab_bg(self.tab_result, "https://images.unsplash.com/photo-1434030216411-0b793f4b4173?w=1280", "result")
        self.setup_tab_bg(self.tab_fees, "https://images.unsplash.com/photo-1554224155-8d04cb21cd6c?w=1280", "fees")
        self.setup_tab_bg(self.tab_exam, "https://images.unsplash.com/photo-1454165804606-c3d57bc86b40?w=1280", "exam")
        self.setup_tab_bg(self.tab_attendance, "https://images.unsplash.com/photo-1509062522246-3755977927d7?w=1280", "attendance")
        self.setup_tab_bg(self.tab_assignment, "https://images.unsplash.com/photo-1488190211105-8b0e65b80b4e?w=1280", "assignment")
        self.setup_tab_bg(self.tab_leave, "https://images.unsplash.com/photo-1517842645767-c639042777db?w=1280", "leave")
        self.setup_tab_bg(self.tab_sports, "https://images.unsplash.com/photo-1461896836934-ffe607ba8211?w=1280", "sports")
        self.setup_tab_bg(self.tab_exams_comp, "https://images.unsplash.com/photo-1516321318423-f06f85e504b3?w=1280", "exams_comp")

        self.setup_dashboard_tab()
        self.setup_student_tab()
        self.setup_result_tab()
        self.setup_fees_tab()
        self.setup_exam_tab()
        self.setup_attendance_tab()
        self.setup_assignment_tab()
        self.setup_leave_tab()
        self.setup_sports_tab()
        self.setup_exams_comp_tab()

    def setup_tab_bg(self, tab, url, name):
        canvas_bg = Canvas(tab, width=1280, height=600, highlightthickness=0)
        canvas_bg.pack(fill=BOTH, expand=True)
        try:
            req = urllib.request.Request(url, headers={'User-Agent': 'Mozilla/5.0'})
            with urllib.request.urlopen(req, timeout=3) as response:
                image_data = response.read()
                pil_img = Image.open(io.BytesIO(image_data)).resize((1280, 600), Image.Resampling.LANCZOS)
                tk_img = ImageTk.PhotoImage(pil_img)
                self.bg_images[name] = tk_img
                canvas_bg.create_image(0, 0, anchor=NW, image=tk_img)
        except Exception:
            canvas_bg.create_rectangle(0, 0, 1280, 600, fill="#eef2f5")

    def go_to_home(self):
        op = messagebox.askyesno("Confirm Logout", "Kiya aap logout karke wapas Login screen par jana chahte hain?", parent=self.root)
        if op:
            self.root.destroy()
            login_root = Tk()
            LoginWindow(login_root)
            login_root.mainloop()

    def create_db(self):
        con = sqlite3.connect(database="student_rms_ultimate.db")
        cur = con.cursor()
        cur.execute("""CREATE TABLE IF NOT EXISTS student(roll TEXT PRIMARY KEY, name TEXT, email TEXT, gender TEXT, dob TEXT, contact TEXT, admission TEXT, semester TEXT, branch TEXT, image_path TEXT, address TEXT)""")
        cur.execute("""CREATE TABLE IF NOT EXISTS result(roll TEXT PRIMARY KEY, name TEXT, course TEXT, marks TEXT, full_marks TEXT, per TEXT, grade TEXT)""")
        cur.execute("""CREATE TABLE IF NOT EXISTS fees(roll TEXT PRIMARY KEY, name TEXT, total_fees TEXT, paid_fees TEXT, dues TEXT, head TEXT, status TEXT)""")
        cur.execute("""CREATE TABLE IF NOT EXISTS exam_form(roll TEXT PRIMARY KEY, name TEXT, course TEXT, semester TEXT, form_type TEXT, subjects TEXT)""")
        cur.execute("""CREATE TABLE IF NOT EXISTS attendance(roll TEXT PRIMARY KEY, name TEXT, total_days TEXT, present_days TEXT, percentage TEXT, status TEXT)""")
        cur.execute("""CREATE TABLE IF NOT EXISTS assignment(roll TEXT PRIMARY KEY, name TEXT, subject TEXT, title TEXT, submission_date TEXT, status TEXT)""")
        cur.execute("""CREATE TABLE IF NOT EXISTS leave_app(roll TEXT PRIMARY KEY, name TEXT, reason TEXT, from_date TEXT, to_date TEXT, status TEXT)""")
        cur.execute("""CREATE TABLE IF NOT EXISTS sports(roll TEXT PRIMARY KEY, name TEXT, game_name TEXT, category TEXT, achievement TEXT)""")
        cur.execute("""CREATE TABLE IF NOT EXISTS qualifying_exams(roll TEXT PRIMARY KEY, name TEXT, exam_name TEXT, score TEXT, year TEXT, status TEXT)""")
        con.commit()
        con.close()

    # 🎨 ATTRACTIVE DASHBOARD MODULE WITH AUTOMATIC SLIDING BACKGROUND IMAGES
    def setup_dashboard_tab(self):
        # Background Canvas for Dashboard Sliding Images
        self.dash_canvas = Canvas(self.tab_dashboard, width=1280, height=600, highlightthickness=0)
        self.dash_canvas.pack(fill=BOTH, expand=True)

        dash_image_urls = [
            "https://images.unsplash.com/photo-1551836022-d5d88e9218df?w=1280",
            "https://images.unsplash.com/photo-1519389950473-47ba0277781c?w=1280",
            "https://images.unsplash.com/photo-1531482615713-2afd69097998?w=1280"
        ]
        
        self.dash_img_index = 0
        self.dash_tk_images = []
        self.dash_canvas.create_rectangle(0, 0, 1280, 600, fill="#0F172A")
        
        for url in dash_image_urls:
            try:
                req = urllib.request.Request(url, headers={'User-Agent': 'Mozilla/5.0'})
                with urllib.request.urlopen(req, timeout=3) as response:
                    image_data = response.read()
                    pil_img = Image.open(io.BytesIO(image_data)).resize((1280, 600), Image.Resampling.LANCZOS)
                    self.dash_tk_images.append(ImageTk.PhotoImage(pil_img))
            except Exception:
                pass
                
        if self.dash_tk_images:
            self.dash_bg_item = self.dash_canvas.create_image(0, 0, anchor=NW, image=self.dash_tk_images[0])
            self.start_dash_slideshow()

        # Header Banner Frame placed over canvas using window item
        header = Frame(self.dash_canvas, bg="#1E293B", height=60)
        self.dash_canvas.create_window(0, 0, anchor=NW, window=header, width=1280, height=60)
        
        Label(header, text="✨ SYSTEM ANALYTICS & OVERVIEW", font=("Helvetica", 14, "bold"), bg="#1E293B", fg="#38BDF8").place(x=30, y=18)
        
        btn_refresh = Button(header, text="🔄 Live Refresh", command=self.update_dashboard_stats, font=("Helvetica", 10, "bold"), bg="#0284C7", fg="white", cursor="hand2", bd=0, activebackground="#0369A1")
        btn_refresh.place(x=1120, y=12, width=130, height=35)

        # Main Content Container Frame over canvas
        content_frame = Frame(self.dash_canvas, bg="#0F172A")
        content_frame.place(x=50, y=80, width=1180, height=480)

        card_configs = [
            ("👨‍🎓 Total Enrolled Students", "0", "#00F2FE", "#4FACFE", 0, 0, lambda: self.notebook.select(1)),
            ("📜 Generated Results", "0", "#11998E", "#38EF7D", 0, 1, lambda: self.notebook.select(2)),
            ("💳 Total Fee Records", "0", "#FF8008", "#FFC837", 0, 2, lambda: self.notebook.select(3)),
            ("📝 Submitted Exam Forms", "0", "#8E2DE2", "#4A00E0", 1, 0, lambda: self.notebook.select(4)),
            ("✉️ Pending Leave Apps", "0", "#FF416C", "#FF4B2B", 1, 1, lambda: self.notebook.select(7)),
            ("📂 Submitted Assignments", "0", "#3A1C71", "#D76D77", 1, 2, lambda: self.notebook.select(5))
        ]

        self.card_labels = {}
        for title, val, c1, c2, r, c, cmd in card_configs:
            lbl_val = self.create_attractive_card(content_frame, title, val, c1, r, c, cmd)
            self.card_labels[title] = lbl_val

        self.update_dashboard_stats()

    def start_dash_slideshow(self):
        if self.dash_tk_images:
            self.dash_img_index = (self.dash_img_index + 1) % len(self.dash_tk_images)
            try:
                self.dash_canvas.itemconfig(self.dash_bg_item, image=self.dash_tk_images[self.dash_img_index])
                self.root.after(3500, self.start_dash_slideshow)
            except Exception:
                pass

    def create_attractive_card(self, parent, title_text, value_text, main_color, row, col, click_cmd):
        card = Frame(parent, bg="#1E293B", bd=0, highlightbackground=main_color, highlightthickness=2, cursor="hand2")
        card.grid(row=row, column=col, padx=20, pady=20, sticky="nsew")
        
        parent.grid_columnconfigure(col, weight=1)
        parent.grid_rowconfigure(row, weight=1)

        color_strip = Frame(card, bg=main_color, height=8)
        color_strip.pack(fill=X, side=TOP)

        lbl_title = Label(card, text=title_text, font=("Helvetica", 11, "bold"), bg="#1E293B", fg="#94A3B8")
        lbl_title.pack(pady=(15, 5), padx=10)
        
        lbl_val = Label(card, text=value_text, font=("Helvetica", 28, "bold"), bg="#1E293B", fg="white")
        lbl_val.pack(pady=(0, 15), padx=10)

        def on_enter(e):
            card.config(bg="#334155")
            lbl_title.config(bg="#334155", fg="#F8FAFC")
            lbl_val.config(bg="#334155")

        def on_leave(e):
            card.config(bg="#1E293B")
            lbl_title.config(bg="#1E293B", fg="#94A3B8")
            lbl_val.config(bg="#1E293B")

        for widget in (card, lbl_title, lbl_val):
            widget.bind("<Button-1>", lambda e: click_cmd())
            widget.bind("<Enter>", on_enter)
            widget.bind("<Leave>", on_leave)

        return lbl_val

    def update_dashboard_stats(self):
        try:
            con = sqlite3.connect(database="student_rms_ultimate.db")
            cur = con.cursor()
            
            cur.execute("SELECT COUNT(*) FROM student")
            self.card_labels["👨‍🎓 Total Enrolled Students"].config(text=str(cur.fetchone()[0]))
            
            cur.execute("SELECT COUNT(*) FROM result")
            self.card_labels["📜 Generated Results"].config(text=str(cur.fetchone()[0]))
            
            cur.execute("SELECT COUNT(*) FROM fees")
            self.card_labels["💳 Total Fee Records"].config(text=str(cur.fetchone()[0]))
            
            cur.execute("SELECT COUNT(*) FROM exam_form")
            self.card_labels["📝 Submitted Exam Forms"].config(text=str(cur.fetchone()[0]))

            cur.execute("SELECT COUNT(*) FROM leave_app WHERE status='Pending Approval'")
            self.card_labels["✉️ Pending Leave Apps"].config(text=str(cur.fetchone()[0]))

            cur.execute("SELECT COUNT(*) FROM assignment")
            self.card_labels["📂 Submitted Assignments"].config(text=str(cur.fetchone()[0]))

            con.close()
        except Exception:
            pass

    # 1. STUDENT TAB
    def setup_student_tab(self):
        title = Label(self.tab_student, text="Student Information & Academic Management", font=("Helvetica", 15, "bold"), bg="#b30000", fg="white")
        title.place(x=0, y=0, relwidth=1, height=40)
        
        self.var_roll = StringVar()
        self.var_name = StringVar()
        self.var_email = StringVar()
        self.var_gender = StringVar()
        self.var_dob = StringVar()
        self.var_contact = StringVar()
        self.var_admission = StringVar()
        self.var_semester = StringVar()
        self.var_branch = StringVar()
        self.var_image = StringVar()
        
        Label(self.tab_student, text="Roll No.", font=("Helvetica", 10, "bold"), bg="white").place(x=10, y=50)
        Entry(self.tab_student, textvariable=self.var_roll, font=("Helvetica", 10), bg="white").place(x=110, y=50, width=160)
        
        Label(self.tab_student, text="Name", font=("Helvetica", 10, "bold"), bg="white").place(x=10, y=90)
        Entry(self.tab_student, textvariable=self.var_name, font=("Helvetica", 10), bg="white").place(x=110, y=90, width=160)
        
        Label(self.tab_student, text="Email", font=("Helvetica", 10, "bold"), bg="white").place(x=10, y=130)
        Entry(self.tab_student, textvariable=self.var_email, font=("Helvetica", 10), bg="white").place(x=110, y=130, width=160)
        
        Label(self.tab_student, text="Gender", font=("Helvetica", 10, "bold"), bg="white").place(x=10, y=170)
        self.txt_gender = ttk.Combobox(self.tab_student, textvariable=self.var_gender, values=("Select", "Male", "Female", "Other"), state="readonly", font=("Helvetica", 10))
        self.txt_gender.place(x=110, y=170, width=160)
        self.txt_gender.current(0)
        
        Label(self.tab_student, text="D.O.B.", font=("Helvetica", 10, "bold"), bg="white").place(x=10, y=210)
        Entry(self.tab_student, textvariable=self.var_dob, font=("Helvetica", 10), bg="white").place(x=110, y=210, width=160)
        
        Label(self.tab_student, text="Contact", font=("Helvetica", 10, "bold"), bg="white").place(x=290, y=50)
        Entry(self.tab_student, textvariable=self.var_contact, font=("Helvetica", 10), bg="white").place(x=380, y=50, width=160)
        
        Label(self.tab_student, text="Admission", font=("Helvetica", 10, "bold"), bg="white").place(x=290, y=90)
        Entry(self.tab_student, textvariable=self.var_admission, font=("Helvetica", 10), bg="white").place(x=380, y=90, width=160)
        
        Label(self.tab_student, text="Semester", font=("Helvetica", 10, "bold"), bg="white").place(x=290, y=130)
        self.txt_sem = ttk.Combobox(self.tab_student, textvariable=self.var_semester, values=("Select", "1st Sem", "2nd Sem", "3rd Sem", "4th Sem", "5th Sem", "6th Sem", "7th Sem", "8th Sem"), state="readonly", font=("Helvetica", 10))
        self.txt_sem.place(x=380, y=130, width=160)
        self.txt_sem.current(0)
        
        Label(self.tab_student, text="Branch", font=("Helvetica", 10, "bold"), bg="white").place(x=290, y=170)
        self.txt_branch = ttk.Combobox(self.tab_student, textvariable=self.var_branch, values=("Select", "BCA", "MCA", "B.Tech CSE", "B.Tech IT", "BBA", "MBA"), state="readonly", font=("Helvetica", 10))
        self.txt_branch.place(x=380, y=170, width=160)
        self.txt_branch.current(0)
        
        Label(self.tab_student, text="Photo File", font=("Helvetica", 10, "bold"), bg="white").place(x=290, y=210)
        Entry(self.tab_student, textvariable=self.var_image, font=("Helvetica", 9), bg="#edf2f7", state="readonly").place(x=380, y=210, width=105)
        Button(self.tab_student, text="Browse", command=self.browse_image, font=("Helvetica", 9, "bold"), bg="#b30000", fg="white", cursor="hand2").place(x=490, y=208, width=50, height=24)
        
        Label(self.tab_student, text="Address", font=("Helvetica", 10, "bold"), bg="white").place(x=560, y=50)
        self.txt_address = Text(self.tab_student, font=("Helvetica", 10), bg="white")
        self.txt_address.place(x=630, y=50, width=220, height=140)

        Label(self.tab_student, text="Photo Preview", font=("Helvetica", 10, "bold"), bg="white").place(x=870, y=50)
        self.lbl_photo_preview = Label(self.tab_student, bg="#d1d5db", bd=2, relief=GROOVE, text="No Image")
        self.lbl_photo_preview.place(x=870, y=75, width=130, height=140)
        
        Button(self.tab_student, text="Save Student", command=self.add_student, font=("Helvetica", 11, "bold"), bg="#b30000", fg="white", cursor="hand2").place(x=110, y=255, width=130, height=35)
        Button(self.tab_student, text="Update", command=self.update_student, font=("Helvetica", 11, "bold"), bg="#2e7d32", fg="white", cursor="hand2").place(x=255, y=255, width=100, height=35)
        Button(self.tab_student, text="Delete", command=self.delete_student, font=("Helvetica", 11, "bold"), bg="#c62828", fg="white", cursor="hand2").place(x=370, y=255, width=100, height=35)
        Button(self.tab_student, text="Clear", command=self.clear_student, font=("Helvetica", 11, "bold"), bg="#546e7a", fg="white", cursor="hand2").place(x=485, y=255, width=100, height=35)
        
        table_frame = Frame(self.tab_student, bd=2, relief=RIDGE)
        table_frame.place(x=10, y=300, width=1255, height=260)
        
        scrolly = Scrollbar(table_frame, orient=VERTICAL)
        scrollx = Scrollbar(table_frame, orient=HORIZONTAL)
        self.student_table = ttk.Treeview(table_frame, columns=("roll", "name", "email", "gender", "dob", "contact", "admission", "semester", "branch", "image"), yscrollcommand=scrolly.set, xscrollcommand=scrollx.set)
        scrollx.pack(side=BOTTOM, fill=X)
        scrolly.pack(side=RIGHT, fill=Y)
        scrollx.config(command=self.student_table.xview)
        scrolly.config(command=self.student_table.yview)
        
        for col, text in [("roll", "Roll No"), ("name", "Name"), ("email", "Email"), ("gender", "Gender"), ("dob", "D.O.B"), ("contact", "Contact"), ("admission", "Admission"), ("semester", "Semester"), ("branch", "Branch"), ("image", "Photo File")]:
            self.student_table.heading(col, text=text)
            self.student_table.column(col, width=120)
        self.student_table["show"] = 'headings'
        self.student_table.pack(fill=BOTH, expand=1)
        self.student_table.bind("<ButtonRelease-1>", self.get_student_data)
        self.show_student_table()

    def browse_image(self):
        filename = filedialog.askopenfilename(initialdir="/", title="Select Student Photo", filetypes=(("Image files", "*.jpg *.jpeg *.png"), ("All files", "*.*")))
        if filename:
            self.var_image.set(filename)
            self.show_photo_preview(filename)

    def show_photo_preview(self, img_path):
        try:
            pil_img = Image.open(img_path).resize((130, 140), Image.Resampling.LANCZOS)
            self.student_photo_img = ImageTk.PhotoImage(pil_img)
            self.lbl_photo_preview.config(image=self.student_photo_img, text="")
        except Exception:
            self.lbl_photo_preview.config(image="", text="No Image")

    def add_student(self):
        if self.var_roll.get() == "":
            messagebox.showerror("Error", "Roll Number required", parent=self.root)
            return
        try:
            con = sqlite3.connect(database="student_rms_ultimate.db")
            cur = con.cursor()
            cur.execute("SELECT * FROM student WHERE roll=?", (self.var_roll.get(),))
            if cur.fetchone() is not None:
                messagebox.showerror("Error", "Roll Number already exists", parent=self.root)
            else:
                cur.execute("INSERT INTO student VALUES(?,?,?,?,?,?,?,?,?,?,?)",
                            (self.var_roll.get(), self.var_name.get(), self.var_email.get(), self.var_gender.get(),
                             self.var_dob.get(), self.var_contact.get(), self.var_admission.get(),
                             self.var_semester.get(), self.var_branch.get(), self.var_image.get(), self.txt_address.get("1.0", END)))
                con.commit()
                messagebox.showinfo("Success", "Student added successfully", parent=self.root)
                self.show_student_table()
                self.clear_student()
                self.update_dashboard_stats()
            con.close()
        except Exception as ex:
            messagebox.showerror("Error", str(ex), parent=self.root)

    def show_student_table(self):
        try:
            con = sqlite3.connect(database="student_rms_ultimate.db")
            cur = con.cursor()
            cur.execute("SELECT roll, name, email, gender, dob, contact, admission, semester, branch, image_path FROM student")
            rows = cur.fetchall()
            self.student_table.delete(*self.student_table.get_children())
            for row in rows:
                self.student_table.insert('', END, values=row)
            con.close()
        except:
            pass

    def get_student_data(self, ev):
        r = self.student_table.focus()
        content = self.student_table.item(r)
        row = content['values']
        if row:
            self.var_roll.set(row[0])
            self.var_name.set(row[1])
            self.var_email.set(row[2])
            self.var_gender.set(row[3])
            self.var_dob.set(row[4])
            self.var_contact.set(row[5])
            self.var_admission.set(row[6])
            self.var_semester.set(row[7])
            self.var_branch.set(row[8])
            self.var_image.set(row[9])
            if row[9]:
                self.show_photo_preview(row[9])
            else:
                self.lbl_photo_preview.config(image="", text="No Image")

    def update_student(self):
        if self.var_roll.get() == "":
            messagebox.showerror("Error", "Select student to update", parent=self.root)
            return
        try:
            con = sqlite3.connect(database="student_rms_ultimate.db")
            cur = con.cursor()
            cur.execute("UPDATE student SET name=?, email=?, gender=?, dob=?, contact=?, admission=?, semester=?, branch=?, image_path=?, address=? WHERE roll=?",
                        (self.var_name.get(), self.var_email.get(), self.var_gender.get(), self.var_dob.get(),
                         self.var_contact.get(), self.var_admission.get(), self.var_semester.get(), self.var_branch.get(),
                         self.var_image.get(), self.txt_address.get("1.0", END), self.var_roll.get()))
            con.commit()
            messagebox.showinfo("Success", "Updated successfully", parent=self.root)
            self.show_student_table()
            con.close()
        except Exception as ex:
            messagebox.showerror("Error", str(ex), parent=self.root)

    def delete_student(self):
        if self.var_roll.get() == "":
            messagebox.showerror("Error", "Select student to delete", parent=self.root)
            return
        try:
            con = sqlite3.connect(database="student_rms_ultimate.db")
            cur = con.cursor()
            cur.execute("DELETE FROM student WHERE roll=?", (self.var_roll.get(),))
            con.commit()
            messagebox.showinfo("Success", "Deleted successfully", parent=self.root)
            self.show_student_table()
            self.clear_student()
            self.update_dashboard_stats()
            con.close()
        except Exception as ex:
            messagebox.showerror("Error", str(ex), parent=self.root)

    def clear_student(self):
        self.var_roll.set("")
        self.var_name.set("")
        self.var_email.set("")
        self.txt_gender.current(0)
        self.var_dob.set("")
        self.var_contact.set("")
        self.var_admission.set("")
        self.txt_sem.current(0)
        self.txt_branch.current(0)
        self.var_image.set("")
        self.txt_address.delete("1.0", END)
        self.lbl_photo_preview.config(image="", text="No Image")

    # 2. RESULT TAB
    def setup_result_tab(self):
        title = Label(self.tab_result, text="Result Generation Portal", font=("Helvetica", 15, "bold"), bg="#b30000", fg="white")
        title.place(x=0, y=0, relwidth=1, height=40)
        
        self.var_res_roll = StringVar()
        self.var_res_name = StringVar()
        self.var_res_course = StringVar()
        self.var_marks = StringVar()
        self.var_full_marks = StringVar()
        self.var_percentage = StringVar()
        self.var_grade = StringVar()
        
        Label(self.tab_result, text="Roll No.", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=70)
        Entry(self.tab_result, textvariable=self.var_res_roll, font=("Helvetica", 11), bg="white").place(x=180, y=70, width=180)
        Button(self.tab_result, text="Fetch", command=self.fetch_res_student, font=("Helvetica", 11, "bold"), bg="#2196f3", fg="white").place(x=370, y=68, width=100, height=30)
        
        Label(self.tab_result, text="Name", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=120)
        Entry(self.tab_result, textvariable=self.var_res_name, font=("Helvetica", 11), bg="#edf2f7", state="readonly").place(x=180, y=120, width=180)
        
        Label(self.tab_result, text="Course", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=170)
        Entry(self.tab_result, textvariable=self.var_res_course, font=("Helvetica", 11), bg="white").place(x=180, y=170, width=180)
        
        Label(self.tab_result, text="Marks Obtained", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=220)
        Entry(self.tab_result, textvariable=self.var_marks, font=("Helvetica", 11), bg="white").place(x=180, y=220, width=180)
        
        Label(self.tab_result, text="Full Marks", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=270)
        Entry(self.tab_result, textvariable=self.var_full_marks, font=("Helvetica", 11), bg="white").place(x=180, y=270, width=180)
        
        Label(self.tab_result, text="Percentage", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=320)
        Entry(self.tab_result, textvariable=self.var_percentage, font=("Helvetica", 11), bg="#edf2f7", state="readonly").place(x=180, y=320, width=180)
        
        Label(self.tab_result, text="Grade", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=370)
        Entry(self.tab_result, textvariable=self.var_grade, font=("Helvetica", 11), bg="#edf2f7", state="readonly").place(x=180, y=370, width=180)
        
        Button(self.tab_result, text="Save Result", command=self.save_result, font=("Helvetica", 11, "bold"), bg="#b30000", fg="white").place(x=180, y=425, width=180, height=40)
        Button(self.tab_result, text="Clear Fields", command=self.clear_result, font=("Helvetica", 11, "bold"), bg="#546e7a", fg="white").place(x=380, y=425, width=180, height=40)

    def fetch_res_student(self):
        try:
            con = sqlite3.connect(database="student_rms_ultimate.db")
            cur = con.cursor()
            cur.execute("SELECT name, branch FROM student WHERE roll=?", (self.var_res_roll.get(),))
            row = cur.fetchone()
            if row:
                self.var_res_name.set(row[0])
                self.var_res_course.set(row[1])
            con.close()
        except: pass

    def save_result(self):
        try:
            marks, full = float(self.var_marks.get()), float(self.var_full_marks.get())
            per = (marks/full)*100
            grade = "A+" if per >= 85 else ("A" if per >= 70 else "B")
            per_str = f"{per:.2f}%"
            self.var_percentage.set(per_str)
            self.var_grade.set(grade)
            con = sqlite3.connect(database="student_rms_ultimate.db")
            cur = con.cursor()
            cur.execute("INSERT OR REPLACE INTO result VALUES(?,?,?,?,?,?,?)", (self.var_res_roll.get(), self.var_res_name.get(), self.var_res_course.get(), str(marks), str(full), per_str, grade))
            con.commit()
            con.close()
            messagebox.showinfo("Success", "Result Saved Successfully!", parent=self.root)
            self.update_dashboard_stats()
        except Exception as e: messagebox.showerror("Error", str(e), parent=self.root)

    def clear_result(self):
        self.var_res_roll.set("")
        self.var_res_name.set("")
        self.var_res_course.set("")
        self.var_marks.set("")
        self.var_full_marks.set("")
        self.var_percentage.set("")
        self.var_grade.set("")

    # 3. FEES TAB
    def setup_fees_tab(self):
        title = Label(self.tab_fees, text="Fees Structure & Tracking", font=("Helvetica", 15, "bold"), bg="#b30000", fg="white")
        title.place(x=0, y=0, relwidth=1, height=40)
        
        self.var_fee_roll = StringVar()
        self.var_fee_name = StringVar()
        self.var_total_fees = StringVar()
        self.var_paid_fees = StringVar()
        self.var_fee_head = StringVar()
        
        Label(self.tab_fees, text="Roll No.", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=70)
        Entry(self.tab_fees, textvariable=self.var_fee_roll, font=("Helvetica", 11), bg="white").place(x=180, y=70, width=180)
        Button(self.tab_fees, text="Fetch", command=self.fetch_fee_student, font=("Helvetica", 11, "bold"), bg="#2196f3", fg="white").place(x=370, y=68, width=100, height=30)
        
        Label(self.tab_fees, text="Name", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=120)
        Entry(self.tab_fees, textvariable=self.var_fee_name, font=("Helvetica", 11), bg="#edf2f7", state="readonly").place(x=180, y=120, width=180)
        
        Label(self.tab_fees, text="Total Fees", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=170)
        Entry(self.tab_fees, textvariable=self.var_total_fees, font=("Helvetica", 11), bg="white").place(x=180, y=170, width=180)
        
        Label(self.tab_fees, text="Paid Amount", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=220)
        Entry(self.tab_fees, textvariable=self.var_paid_fees, font=("Helvetica", 11), bg="white").place(x=180, y=220, width=180)
        
        Label(self.tab_fees, text="Fee Head", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=270)
        self.txt_fee_head = ttk.Combobox(self.tab_fees, textvariable=self.var_fee_head, values=("Tuition Fees", "Exam Fees", "Hostel Fees", "Library"), state="readonly", font=("Helvetica", 11))
        self.txt_fee_head.place(x=180, y=270, width=180)
        self.txt_fee_head.current(0)
        
        Button(self.tab_fees, text="Save Fees", command=self.save_fees, font=("Helvetica", 11, "bold"), bg="#b30000", fg="white").place(x=180, y=330, width=180, height=40)
        
        fees_frame = Frame(self.tab_fees, bd=2, relief=RIDGE)
        fees_frame.place(x=500, y=70, width=755, height=300)
        self.fees_table = ttk.Treeview(fees_frame, columns=("roll", "name", "total", "paid", "dues", "head", "status"))
        for col, text in [("roll", "Roll No"), ("name", "Name"), ("total", "Total"), ("paid", "Paid"), ("dues", "Dues"), ("head", "Head"), ("status", "Status")]:
            self.fees_table.heading(col, text=text)
            self.fees_table.column(col, width=105)
        self.fees_table["show"] = 'headings'
        self.fees_table.pack(fill=BOTH, expand=1)
        self.show_fees_table()

    def fetch_fee_student(self):
        try:
            con = sqlite3.connect(database="student_rms_ultimate.db")
            cur = con.cursor()
            cur.execute("SELECT name FROM student WHERE roll=?", (self.var_fee_roll.get(),))
            row = cur.fetchone()
            if row: self.var_fee_name.set(row[0])
            con.close()
        except: pass

    def save_fees(self):
        try:
            total, paid = float(self.var_total_fees.get()), float(self.var_paid_fees.get())
            dues = total - paid
            status = "Paid" if dues <= 0 else "Pending"
            con = sqlite3.connect(database="student_rms_ultimate.db")
            cur = con.cursor()
            cur.execute("INSERT OR REPLACE INTO fees VALUES(?,?,?,?,?,?,?)", (self.var_fee_roll.get(), self.var_fee_name.get(), str(total), str(paid), str(dues), self.var_fee_head.get(), status))
            con.commit()
            con.close()
            messagebox.showinfo("Success", f"Fees Saved! Dues: {dues}", parent=self.root)
            self.show_fees_table()
            self.update_dashboard_stats()
        except Exception as e: messagebox.showerror("Error", str(e), parent=self.root)

    def show_fees_table(self):
        try:
            con = sqlite3.connect(database="student_rms_ultimate.db")
            cur = con.cursor()
            cur.execute("SELECT * FROM fees")
            rows = cur.fetchall()
            self.fees_table.delete(*self.fees_table.get_children())
            for row in rows: self.fees_table.insert('', END, values=row)
            con.close()
        except: pass

    # 4. EXAM FORM TAB
    def setup_exam_tab(self):
        title = Label(self.tab_exam, text="Examination Form Portal", font=("Helvetica", 15, "bold"), bg="#b30000", fg="white")
        title.place(x=0, y=0, relwidth=1, height=40)
        
        self.var_ex_roll = StringVar()
        self.var_ex_name = StringVar()
        self.var_ex_course = StringVar()
        self.var_ex_sem = StringVar()
        self.var_ex_formtype = StringVar()
        self.var_ex_subs = StringVar()
        
        Label(self.tab_exam, text="Roll No.", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=60)
        Entry(self.tab_exam, textvariable=self.var_ex_roll, font=("Helvetica", 11), bg="white").place(x=180, y=60, width=180)
        Button(self.tab_exam, text="Fetch", command=self.fetch_exam_student, font=("Helvetica", 11, "bold"), bg="#2196f3", fg="white").place(x=370, y=58, width=100, height=30)
        
        Label(self.tab_exam, text="Name", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=105)
        Entry(self.tab_exam, textvariable=self.var_ex_name, font=("Helvetica", 11), bg="#edf2f7", state="readonly").place(x=180, y=105, width=180)
        
        Label(self.tab_exam, text="Course", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=150)
        Entry(self.tab_exam, textvariable=self.var_ex_course, font=("Helvetica", 11), bg="white").place(x=180, y=150, width=180)
        
        Label(self.tab_exam, text="Semester", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=195)
        Entry(self.tab_exam, textvariable=self.var_ex_sem, font=("Helvetica", 11), bg="white").place(x=180, y=195, width=180)
        
        Label(self.tab_exam, text="Form Type", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=240)
        self.cmb_formtype = ttk.Combobox(self.tab_exam, textvariable=self.var_ex_formtype, values=("Semester Form", "Reappear Form"), state="readonly", font=("Helvetica", 11))
        self.cmb_formtype.place(x=180, y=240, width=180)
        self.cmb_formtype.current(0)
        
        Label(self.tab_exam, text="All Subjects\n(Comma Separated)", font=("Helvetica", 10, "bold"), bg="white", justify=LEFT).place(x=30, y=285)
        Entry(self.tab_exam, textvariable=self.var_ex_subs, font=("Helvetica", 11), bg="white").place(x=180, y=290, width=250)
        
        Button(self.tab_exam, text="Submit Form", command=self.save_exam_form, font=("Helvetica", 11, "bold"), bg="#b30000", fg="white", cursor="hand2").place(x=180, y=340, width=180, height=40)
        
        ex_frame = Frame(self.tab_exam, bd=2, relief=RIDGE)
        ex_frame.place(x=450, y=60, width=800, height=320)
        self.ex_table = ttk.Treeview(ex_frame, columns=("roll", "name", "course", "sem", "formtype", "subs"))
        for col, text in [("roll", "Roll No"), ("name", "Name"), ("course", "Course"), ("sem", "Sem"), ("formtype", "Form Type"), ("subs", "Submitted Subjects")]:
            self.ex_table.heading(col, text=text)
            self.ex_table.column(col, width=130)
        self.ex_table["show"] = 'headings'
        self.ex_table.pack(fill=BOTH, expand=1)
        self.show_exam_table()

    def fetch_exam_student(self):
        try:
            con = sqlite3.connect(database="student_rms_ultimate.db")
            cur = con.cursor()
            cur.execute("SELECT name, branch, semester FROM student WHERE roll=?", (self.var_ex_roll.get(),))
            row = cur.fetchone()
            if row:
                self.var_ex_name.set(row[0])
                self.var_ex_course.set(row[1])
                self.var_ex_sem.set(row[2])
            con.close()
        except Exception as e:
            messagebox.showerror("Error", str(e), parent=self.root)

    def save_exam_form(self):
        if self.var_ex_roll.get() == "" or self.var_ex_subs.get() == "":
            messagebox.showerror("Error", "Please fill Roll No and Subjects", parent=self.root)
            return
        try:
            con = sqlite3.connect(database="student_rms_ultimate.db")
            cur = con.cursor()
            cur.execute("INSERT OR REPLACE INTO exam_form VALUES(?,?,?,?,?,?)", 
                        (self.var_ex_roll.get(), self.var_ex_name.get(), self.var_ex_course.get(), self.var_ex_sem.get(), self.var_ex_formtype.get(), self.var_ex_subs.get()))
            con.commit()
            con.close()
            messagebox.showinfo("Success", f"{self.var_ex_formtype.get()} Submitted Successfully!", parent=self.root)
            self.show_exam_table()
            self.update_dashboard_stats()
        except Exception as e: 
            messagebox.showerror("Error", str(e), parent=self.root)

    def show_exam_table(self):
        try:
            con = sqlite3.connect(database="student_rms_ultimate.db")
            cur = con.cursor()
            cur.execute("SELECT * FROM exam_form")
            rows = cur.fetchall()
            self.ex_table.delete(*self.ex_table.get_children())
            for row in rows: 
                self.ex_table.insert('', END, values=row)
            con.close()
        except Exception as e:
            pass

    # 5. ATTENDANCE TAB
    def setup_attendance_tab(self):
        title = Label(self.tab_attendance, text="Student Attendance Management", font=("Helvetica", 15, "bold"), bg="#b30000", fg="white")
        title.place(x=0, y=0, relwidth=1, height=40)
        
        self.var_att_roll = StringVar()
        self.var_att_name = StringVar()
        self.var_att_total = StringVar()
        self.var_att_present = StringVar()
        
        Label(self.tab_attendance, text="Roll No.", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=70)
        Entry(self.tab_attendance, textvariable=self.var_att_roll, font=("Helvetica", 11), bg="white").place(x=180, y=70, width=180)
        Button(self.tab_attendance, text="Fetch", command=self.fetch_att_student, font=("Helvetica", 11, "bold"), bg="#2196f3", fg="white").place(x=370, y=68, width=100, height=30)
        
        Label(self.tab_attendance, text="Name", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=120)
        Entry(self.tab_attendance, textvariable=self.var_att_name, font=("Helvetica", 11), bg="#edf2f7", state="readonly").place(x=180, y=120, width=180)
        
        Label(self.tab_attendance, text="Total Classes", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=170)
        Entry(self.tab_attendance, textvariable=self.var_att_total, font=("Helvetica", 11), bg="white").place(x=180, y=170, width=180)
        
        Label(self.tab_attendance, text="Present Days", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=220)
        Entry(self.tab_attendance, textvariable=self.var_att_present, font=("Helvetica", 11), bg="white").place(x=180, y=220, width=180)
        
        Button(self.tab_attendance, text="Save Attendance", command=self.save_attendance, font=("Helvetica", 11, "bold"), bg="#b30000", fg="white").place(x=180, y=280, width=180, height=40)
        
        att_frame = Frame(self.tab_attendance, bd=2, relief=RIDGE)
        att_frame.place(x=450, y=70, width=800, height=300)
        self.att_table = ttk.Treeview(att_frame, columns=("roll", "name", "total", "present", "per", "status"))
        for col, text in [("roll", "Roll No"), ("name", "Name"), ("total", "Total Classes"), ("present", "Present Days"), ("per", "Percentage"), ("status", "Status")]:
            self.att_table.heading(col, text=text)
            self.att_table.column(col, width=130)
        self.att_table["show"] = 'headings'
        self.att_table.pack(fill=BOTH, expand=1)
        self.show_attendance_table()

    def fetch_att_student(self):
        try:
            con = sqlite3.connect(database="student_rms_ultimate.db")
            cur = con.cursor()
            cur.execute("SELECT name FROM student WHERE roll=?", (self.var_att_roll.get(),))
            row = cur.fetchone()
            if row: self.var_att_name.set(row[0])
            con.close()
        except: pass

    def save_attendance(self):
        try:
            total, present = float(self.var_att_total.get()), float(self.var_att_present.get())
            per = (present / total) * 100
            per_str = f"{per:.2f}%"
            status = "Good" if per >= 75 else "Short Attendance"
            con = sqlite3.connect(database="student_rms_ultimate.db")
            cur = con.cursor()
            cur.execute("INSERT OR REPLACE INTO attendance VALUES(?,?,?,?,?,?)", (self.var_att_roll.get(), self.var_att_name.get(), str(total), str(present), per_str, status))
            con.commit()
            con.close()
            messagebox.showinfo("Success", f"Attendance Saved! ({per_str})", parent=self.root)
            self.show_attendance_table()
        except Exception as e: messagebox.showerror("Error", str(e), parent=self.root)

    def show_attendance_table(self):
        try:
            con = sqlite3.connect(database="student_rms_ultimate.db")
            cur = con.cursor()
            cur.execute("SELECT * FROM attendance")
            rows = cur.fetchall()
            self.att_table.delete(*self.att_table.get_children())
            for row in rows: self.att_table.insert('', END, values=row)
            con.close()
        except: pass

    # 6. ASSIGNMENTS TAB
    def setup_assignment_tab(self):
        title = Label(self.tab_assignment, text="Assignment Submission Portal", font=("Helvetica", 15, "bold"), bg="#b30000", fg="white")
        title.place(x=0, y=0, relwidth=1, height=40)
        
        self.var_asg_roll = StringVar()
        self.var_asg_name = StringVar()
        self.var_asg_sub = StringVar()
        self.var_asg_title = StringVar()
        self.var_asg_date = StringVar()
        
        Label(self.tab_assignment, text="Roll No.", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=70)
        Entry(self.tab_assignment, textvariable=self.var_asg_roll, font=("Helvetica", 11), bg="white").place(x=180, y=70, width=180)
        Button(self.tab_assignment, text="Fetch", command=self.fetch_asg_student, font=("Helvetica", 11, "bold"), bg="#2196f3", fg="white").place(x=370, y=68, width=100, height=30)
        
        Label(self.tab_assignment, text="Name", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=120)
        Entry(self.tab_assignment, textvariable=self.var_asg_name, font=("Helvetica", 11), bg="#edf2f7", state="readonly").place(x=180, y=120, width=180)
        
        Label(self.tab_assignment, text="Subject", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=170)
        Entry(self.tab_assignment, textvariable=self.var_asg_sub, font=("Helvetica", 11), bg="white").place(x=180, y=170, width=180)
        
        Label(self.tab_assignment, text="Assignment Title", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=220)
        Entry(self.tab_assignment, textvariable=self.var_asg_title, font=("Helvetica", 11), bg="white").place(x=180, y=220, width=180)
        
        Label(self.tab_assignment, text="Submission Date", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=270)
        Entry(self.tab_assignment, textvariable=self.var_asg_date, font=("Helvetica", 11), bg="white").place(x=180, y=270, width=180)
        
        Button(self.tab_assignment, text="Submit Assignment", command=self.save_assignment, font=("Helvetica", 11, "bold"), bg="#b30000", fg="white").place(x=180, y=330, width=180, height=40)
        
        asg_frame = Frame(self.tab_assignment, bd=2, relief=RIDGE)
        asg_frame.place(x=450, y=70, width=800, height=300)
        self.asg_table = ttk.Treeview(asg_frame, columns=("roll", "name", "sub", "title", "date", "status"))
        for col, text in [("roll", "Roll No"), ("name", "Name"), ("sub", "Subject"), ("title", "Title"), ("date", "Date"), ("status", "Status")]:
            self.asg_table.heading(col, text=text)
            self.asg_table.column(col, width=130)
        self.asg_table["show"] = 'headings'
        self.asg_table.pack(fill=BOTH, expand=1)
        self.show_assignment_table()

    def fetch_asg_student(self):
        try:
            con = sqlite3.connect(database="student_rms_ultimate.db")
            cur = con.cursor()
            cur.execute("SELECT name FROM student WHERE roll=?", (self.var_asg_roll.get(),))
            row = cur.fetchone()
            if row: self.var_asg_name.set(row[0])
            con.close()
        except: pass

    def save_assignment(self):
        try:
            con = sqlite3.connect(database="student_rms_ultimate.db")
            cur = con.cursor()
            cur.execute("INSERT OR REPLACE INTO assignment VALUES(?,?,?,?,?,?)", (self.var_asg_roll.get(), self.var_asg_name.get(), self.var_asg_sub.get(), self.var_asg_title.get(), self.var_asg_date.get(), "Submitted"))
            con.commit()
            con.close()
            messagebox.showinfo("Success", "Assignment Submitted!", parent=self.root)
            self.show_assignment_table()
            self.update_dashboard_stats()
        except Exception as e: messagebox.showerror("Error", str(e), parent=self.root)

    def show_assignment_table(self):
        try:
            con = sqlite3.connect(database="student_rms_ultimate.db")
            cur = con.cursor()
            cur.execute("SELECT * FROM assignment")
            rows = cur.fetchall()
            self.asg_table.delete(*self.asg_table.get_children())
            for row in rows: self.asg_table.insert('', END, values=row)
            con.close()
        except: pass

    # 7. LEAVE APPLICATION TAB
    def setup_leave_tab(self):
        title = Label(self.tab_leave, text="Online Leave Application Portal", font=("Helvetica", 15, "bold"), bg="#b30000", fg="white")
        title.place(x=0, y=0, relwidth=1, height=40)
        
        self.var_lev_roll = StringVar()
        self.var_lev_name = StringVar()
        self.var_lev_reason = StringVar()
        self.var_lev_from = StringVar()
        self.var_lev_to = StringVar()
        
        Label(self.tab_leave, text="Roll No.", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=70)
        Entry(self.tab_leave, textvariable=self.var_lev_roll, font=("Helvetica", 11), bg="white").place(x=180, y=70, width=180)
        Button(self.tab_leave, text="Fetch", command=self.fetch_leave_student, font=("Helvetica", 11, "bold"), bg="#2196f3", fg="white").place(x=370, y=68, width=100, height=30)
        
        Label(self.tab_leave, text="Name", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=120)
        Entry(self.tab_leave, textvariable=self.var_lev_name, font=("Helvetica", 11), bg="#edf2f7", state="readonly").place(x=180, y=120, width=180)
        
        Label(self.tab_leave, text="Reason", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=170)
        Entry(self.tab_leave, textvariable=self.var_lev_reason, font=("Helvetica", 11), bg="white").place(x=180, y=170, width=180)
        
        Label(self.tab_leave, text="From Date", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=220)
        Entry(self.tab_leave, textvariable=self.var_lev_from, font=("Helvetica", 11), bg="white").place(x=180, y=220, width=180)
        
        Label(self.tab_leave, text="To Date", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=270)
        Entry(self.tab_leave, textvariable=self.var_lev_to, font=("Helvetica", 11), bg="white").place(x=180, y=270, width=180)
        
        Button(self.tab_leave, text="Apply Leave", command=self.save_leave, font=("Helvetica", 11, "bold"), bg="#b30000", fg="white", cursor="hand2").place(x=180, y=330, width=130, height=40)
        Button(self.tab_leave, text="✔ Approve Leave", command=self.approve_leave, font=("Helvetica", 11, "bold"), bg="#2e7d32", fg="white", cursor="hand2").place(x=320, y=330, width=150, height=40)
        
        lev_frame = Frame(self.tab_leave, bd=2, relief=RIDGE)
        lev_frame.place(x=490, y=70, width=760, height=300)
        self.lev_table = ttk.Treeview(lev_frame, columns=("roll", "name", "reason", "from", "to", "status"))
        for col, text in [("roll", "Roll No"), ("name", "Name"), ("reason", "Reason"), ("from", "From"), ("to", "To"), ("status", "Status")]:
            self.lev_table.heading(col, text=text)
            self.lev_table.column(col, width=120)
        self.lev_table["show"] = 'headings'
        self.lev_table.pack(fill=BOTH, expand=1)
        self.lev_table.bind("<ButtonRelease-1>", self.get_leave_data)
        self.show_leave_table()

    def fetch_leave_student(self):
        try:
            con = sqlite3.connect(database="student_rms_ultimate.db")
            cur = con.cursor()
            cur.execute("SELECT name FROM student WHERE roll=?", (self.var_lev_roll.get(),))
            row = cur.fetchone()
            if row: self.var_lev_name.set(row[0])
            con.close()
        except: pass

    def save_leave(self):
        try:
            con = sqlite3.connect(database="student_rms_ultimate.db")
            cur = con.cursor()
            cur.execute("INSERT OR REPLACE INTO leave_app VALUES(?,?,?,?,?,?)", (self.var_lev_roll.get(), self.var_lev_name.get(), self.var_lev_reason.get(), self.var_lev_from.get(), self.var_lev_to.get(), "Pending Approval"))
            con.commit()
            con.close()
            messagebox.showinfo("Success", "Leave Application Submitted!", parent=self.root)
            self.show_leave_table()
            self.update_dashboard_stats()
        except Exception as e: messagebox.showerror("Error", str(e), parent=self.root)

    def approve_leave(self):
        if self.var_lev_roll.get() == "":
            messagebox.showerror("Error", "Please select a Leave Application or enter Roll No.", parent=self.root)
            return
        try:
            con = sqlite3.connect(database="student_rms_ultimate.db")
            cur = con.cursor()
            cur.execute("UPDATE leave_app SET status='Approved' WHERE roll=?", (self.var_lev_roll.get(),))
            con.commit()
            con.close()
            messagebox.showinfo("Success", "Leave Approved Successfully!", parent=self.root)
            self.show_leave_table()
            self.update_dashboard_stats()
        except Exception as e: messagebox.showerror("Error", str(e), parent=self.root)

    def get_leave_data(self, ev):
        r = self.lev_table.focus()
        content = self.lev_table.item(r)
        row = content['values']
        if row:
            self.var_lev_roll.set(row[0])
            self.var_lev_name.set(row[1])
            self.var_lev_reason.set(row[2])
            self.var_lev_from.set(row[3])
            self.var_lev_to.set(row[4])

    def show_leave_table(self):
        try:
            con = sqlite3.connect(database="student_rms_ultimate.db")
            cur = con.cursor()
            cur.execute("SELECT * FROM leave_app")
            rows = cur.fetchall()
            self.lev_table.delete(*self.lev_table.get_children())
            for row in rows: self.lev_table.insert('', END, values=row)
            con.close()
        except: pass

    # 8. SPORTS & GAMES TAB
    def setup_sports_tab(self):
        title = Label(self.tab_sports, text="Sports & Games Portal", font=("Helvetica", 15, "bold"), bg="#b30000", fg="white")
        title.place(x=0, y=0, relwidth=1, height=40)
        
        self.var_sp_roll = StringVar()
        self.var_sp_name = StringVar()
        self.var_sp_game = StringVar()
        self.var_sp_cat = StringVar()
        self.var_sp_ach = StringVar()
        
        Label(self.tab_sports, text="Roll No.", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=70)
        Entry(self.tab_sports, textvariable=self.var_sp_roll, font=("Helvetica", 11), bg="white").place(x=180, y=70, width=180)
        Button(self.tab_sports, text="Fetch", command=self.fetch_sports_student, font=("Helvetica", 11, "bold"), bg="#2196f3", fg="white").place(x=370, y=68, width=100, height=30)
        
        Label(self.tab_sports, text="Name", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=120)
        Entry(self.tab_sports, textvariable=self.var_sp_name, font=("Helvetica", 11), bg="#edf2f7", state="readonly").place(x=180, y=120, width=180)
        
        Label(self.tab_sports, text="Game Name", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=170)
        Entry(self.tab_sports, textvariable=self.var_sp_game, font=("Helvetica", 11), bg="white").place(x=180, y=170, width=180)
        
        Label(self.tab_sports, text="Category", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=220)
        self.txt_sp_cat = ttk.Combobox(self.tab_sports, textvariable=self.var_sp_cat, values=("Cricket", "Football", "Badminton", "Basketball", "Chess", "Athletics"), state="readonly", font=("Helvetica", 11))
        self.txt_sp_cat.place(x=180, y=220, width=180)
        self.txt_sp_cat.current(0)
        
        Label(self.tab_sports, text="Achievement", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=270)
        Entry(self.tab_sports, textvariable=self.var_sp_ach, font=("Helvetica", 11), bg="white").place(x=180, y=270, width=180)
        
        Button(self.tab_sports, text="Save Sports Info", command=self.save_sports, font=("Helvetica", 11, "bold"), bg="#b30000", fg="white").place(x=180, y=330, width=180, height=40)
        
        sp_frame = Frame(self.tab_sports, bd=2, relief=RIDGE)
        sp_frame.place(x=450, y=70, width=800, height=300)
        self.sp_table = ttk.Treeview(sp_frame, columns=("roll", "name", "game", "cat", "ach"))
        for col, text in [("roll", "Roll No"), ("name", "Name"), ("game", "Game"), ("cat", "Category"), ("ach", "Achievement")]:
            self.sp_table.heading(col, text=text)
            self.sp_table.column(col, width=150)
        self.sp_table["show"] = 'headings'
        self.sp_table.pack(fill=BOTH, expand=1)
        self.show_sports_table()

    def fetch_sports_student(self):
        try:
            con = sqlite3.connect(database="student_rms_ultimate.db")
            cur = con.cursor()
            cur.execute("SELECT name FROM student WHERE roll=?", (self.var_sp_roll.get(),))
            row = cur.fetchone()
            if row: self.var_sp_name.set(row[0])
            con.close()
        except: pass

    def save_sports(self):
        try:
            con = sqlite3.connect(database="student_rms_ultimate.db")
            cur = con.cursor()
            cur.execute("INSERT OR REPLACE INTO sports VALUES(?,?,?,?,?)", (self.var_sp_roll.get(), self.var_sp_name.get(), self.var_sp_game.get(), self.var_sp_cat.get(), self.var_sp_ach.get()))
            con.commit()
            con.close()
            messagebox.showinfo("Success", "Sports Info Saved!", parent=self.root)
            self.show_sports_table()
        except Exception as e: messagebox.showerror("Error", str(e), parent=self.root)

    def show_sports_table(self):
        try:
            con = sqlite3.connect(database="student_rms_ultimate.db")
            cur = con.cursor()
            cur.execute("SELECT * FROM sports")
            rows = cur.fetchall()
            self.sp_table.delete(*self.sp_table.get_children())
            for row in rows: self.sp_table.insert('', END, values=row)
            con.close()
        except: pass

    # 9. QUALIFYING EXAMS TAB
    def setup_exams_comp_tab(self):
        title = Label(self.tab_exams_comp, text="Qualifying Exams & Competitions", font=("Helvetica", 15, "bold"), bg="#b30000", fg="white")
        title.place(x=0, y=0, relwidth=1, height=40)
        
        self.var_qc_roll = StringVar()
        self.var_qc_name = StringVar()
        self.var_qc_exam = StringVar()
        self.var_qc_score = StringVar()
        self.var_qc_year = StringVar()
        
        Label(self.tab_exams_comp, text="Roll No.", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=70)
        Entry(self.tab_exams_comp, textvariable=self.var_qc_roll, font=("Helvetica", 11), bg="white").place(x=180, y=70, width=180)
        Button(self.tab_exams_comp, text="Fetch", command=self.fetch_qc_student, font=("Helvetica", 11, "bold"), bg="#2196f3", fg="white").place(x=370, y=68, width=100, height=30)
        
        Label(self.tab_exams_comp, text="Name", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=120)
        Entry(self.tab_exams_comp, textvariable=self.var_qc_name, font=("Helvetica", 11), bg="#edf2f7", state="readonly").place(x=180, y=120, width=180)
        
        Label(self.tab_exams_comp, text="Exam / Comp Name", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=170)
        Entry(self.tab_exams_comp, textvariable=self.var_qc_exam, font=("Helvetica", 11), bg="white").place(x=180, y=170, width=180)
        
        Label(self.tab_exams_comp, text="Score / Rank", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=220)
        Entry(self.tab_exams_comp, textvariable=self.var_qc_score, font=("Helvetica", 11), bg="white").place(x=180, y=220, width=180)
        
        Label(self.tab_exams_comp, text="Year", font=("Helvetica", 11, "bold"), bg="white").place(x=50, y=270)
        Entry(self.tab_exams_comp, textvariable=self.var_qc_year, font=("Helvetica", 11), bg="white").place(x=180, y=270, width=180)
        
        Button(self.tab_exams_comp, text="Save Record", command=self.save_qc, font=("Helvetica", 11, "bold"), bg="#b30000", fg="white").place(x=180, y=330, width=180, height=40)
        
        qc_frame = Frame(self.tab_exams_comp, bd=2, relief=RIDGE)
        qc_frame.place(x=450, y=70, width=800, height=300)
        self.qc_table = ttk.Treeview(qc_frame, columns=("roll", "name", "exam", "score", "year", "status"))
        for col, text in [("roll", "Roll No"), ("name", "Name"), ("exam", "Exam/Competition"), ("score", "Score/Rank"), ("year", "Year"), ("status", "Status")]:
            self.qc_table.heading(col, text=text)
            self.qc_table.column(col, width=130)
        self.qc_table["show"] = 'headings'
        self.qc_table.pack(fill=BOTH, expand=1)
        self.show_qc_table()

    def fetch_qc_student(self):
        try:
            con = sqlite3.connect(database="student_rms_ultimate.db")
            cur = con.cursor()
            cur.execute("SELECT name FROM student WHERE roll=?", (self.var_qc_roll.get(),))
            row = cur.fetchone()
            if row: self.var_qc_name.set(row[0])
            con.close()
        except: pass

    def save_qc(self):
        try:
            con = sqlite3.connect(database="student_rms_ultimate.db")
            cur = con.cursor()
            cur.execute("INSERT OR REPLACE INTO qualifying_exams VALUES(?,?,?,?,?,?)", (self.var_qc_roll.get(), self.var_qc_name.get(), self.var_qc_exam.get(), self.var_qc_score.get(), self.var_qc_year.get(), "Qualified"))
            con.commit()
            con.close()
            messagebox.showinfo("Success", "Qualifying Exam Record Saved!", parent=self.root)
            self.show_qc_table()
        except Exception as e: messagebox.showerror("Error", str(e), parent=self.root)

    def show_qc_table(self):
        try:
            con = sqlite3.connect(database="student_rms_ultimate.db")
            cur = con.cursor()
            cur.execute("SELECT * FROM qualifying_exams")
            rows = cur.fetchall()
            self.qc_table.delete(*self.qc_table.get_children())
            for row in rows: self.qc_table.insert('', END, values=row)
            con.close()
        except: pass


if __name__ == "__main__":
    root = Tk()
    obj = LoginWindow(root)
    root.mainloop()
```
